```python
import numpy as np

# Create two matrices
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix A:\n", A)
print("Matrix B:\n", B)

# Matrix Addition
print("\nAddition:\n", A + B)

# Matrix Subtraction
print("\nSubtraction:\n", A - B)

# Matrix Multiplication
print("\nMatrix Multiplication:\n", A @ B)

# Transpose of matrices
print("\nTranspose of Matrix A:\n", A.T)
print("\nTranspose of Matrix B:\n", B.T)

# Short explanation
print("\nMatrix multiplication multiplies rows of A by columns of B.")
```
